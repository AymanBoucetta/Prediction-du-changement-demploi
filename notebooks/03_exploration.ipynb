```python
import pandas as pd   
df=pd.read_csv("../data/raw/aug_train.csv")
```

```python
# Les cases vides de company_size et company_type tombent-elles sur les mêmes lignes ?
print(df[["company_size", "company_type"]].isna().value_counts())
```

```python
# D'où viennent les cases vides de major_discipline ?
print(df[df["major_discipline"].isna()]["education_level"].value_counts(dropna=False))
```
