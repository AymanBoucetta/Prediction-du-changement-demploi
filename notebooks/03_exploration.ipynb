```python
import pandas as pd   
df=pd.read_csv("../data/raw/aug_train.csv")
```

```python
# Les cases vides de company_size et company_type tombent-elles sur les mêmes lignes ?
print(df[["company_size", "company_type"]].isna().value_counts())
```

```python
# D'où viennent les cases vides de major_discipline ?
print(df[df["major_discipline"].isna()]["education_level"].value_counts(dropna=False))
```

```python
for col in df.select_dtypes(exclude="number").columns:
    print(col)
    print(df[col].value_counts(dropna=False).head(121))
    print()
```

```python
# 1. Les candidats sans taille d'entreprise ont-ils un point commun ? (hypothèse « sans employeur »)
print(df[df["company_size"].isna()]["last_new_job"].value_counts(dropna=False))
print(df[df["company_size"].isna()]["relevent_experience"].value_counts(dropna=False))

# 2. Les candidats sans niveau d'études ont-ils sauté d'autres champs ?
print(df[df["education_level"].isna()].isna().sum())

# 3. Valeurs déguisées dans les colonnes qu'on n'a vues qu'en partie
print(sorted(df["experience"].dropna().unique()))
print(df["city"].str.startswith("city_").all())
```
