```python
import pickle

with open("mes_valeurs.pkl", "rb") as f:
    data = pickle.load(f)

X_train = data["X_train"]
X_test = data["X_test"]
y_train = data["y_train"]
y_test = data["y_test"]
Xtr_prep=data["Xtr_prep"]
Xte_prep=data["Xte_prep"]
```

```python
from sklearn.dummy import DummyClassifier

dummy = DummyClassifier(strategy="most_frequent")
dummy.fit(Xtr_prep, y_train)

print(dummy.score(Xtr_prep, y_train))
print(dummy.predict(Xtr_prep)[:10])
print("Accuracy Train :", dummy.score(Xtr_prep, y_train))
print("Accuracy Test :", dummy.score(Xte_prep, y_test))
```
