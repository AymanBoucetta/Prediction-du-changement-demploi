```python
# Charge les données d'entraînement depuis le fichier CSV.
import pandas as pd
df=pd.read_csv("../data/raw/aug_train.csv")
```

```python
# Sépare les variables explicatives de la cible et retire l'identifiant, qui n'est pas une caractéristique utile.
X = df.drop(columns=["target", "enrollee_id"])
y = df["target"].astype(int)

print(X.shape, y.shape)
print(y.value_counts())
```

```python
# Sépare les données en entraînement et test en conservant la proportion de chaque classe.
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

print(X_train.shape, X_test.shape)
print(y_train.mean(), y_test.mean())
```

```python
# 1. Aucun candidat dans les deux jeux à la fois
print(len(set(X_train.index) & set(X_test.index)))

# 2. Nombre de candidats qui cherchent à changer dans chaque jeu
print(y_train.sum(), y_test.sum())

# 3. Les cases vides sont-elles réparties de façon similaire ?
print(pd.DataFrame({
    "train": X_train.isna().mean().round(3),
    "test": X_test.isna().mean().round(3),
}))
```

```python
# Chaque ville a-t-elle un seul indice de développement ?
print(df.groupby("city")["city_development_index"].nunique().max())

# Les heures de formation des deux groupes, en moyenne
print(df.groupby("target")["training_hours"].mean())
```
