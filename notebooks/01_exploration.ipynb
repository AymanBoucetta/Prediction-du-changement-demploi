```python
# Construit un tableau recapitulatif des types, valeurs manquantes, modalites, exemples et descriptions des colonnes.
summary = pd.DataFrame({
    "type": df.dtypes.astype(str),
    "manquantes_%": (df.isna().mean() * 100).round(1),
    "valeurs_uniques": df.nunique(),
    "exemples": pd.Series({c: df[c].dropna().unique()[:3].tolist() for c in df.columns}),
})
descriptions = {
    "enrollee_id": "identifiant unique du candidat",
    "city": "code de la ville du candidat",
    "city_development_index": "indice de développement de la ville, entre 0 et 1",
    "gender": "	genre du candidat",
    "relevent_experience": "a-t-il une expérience pertinente pour le poste, oui ou non",
    "enrolled_university": "type de cursus universitaire suivi en ce moment, s’il y en a un",
    "education_level": "niveau d’études",
    "major_discipline": "domaine d’études principal",
    "experience": "années d’expérience totale",
    "company_size": "taille de l’entreprise actuelle",
    "company_type": "type de l’entreprise actuelle",
    "last_new_job": "années écoulées entre l’avant-dernier emploi et l’emploi actuel",
    "training_hours": "	nombre d’heures de formation suivies",
    "target": "1 si le candidat cherche à changer d’emploi, 0 sinon",
}
summary["description"] = pd.Series(descriptions)
summary
```
